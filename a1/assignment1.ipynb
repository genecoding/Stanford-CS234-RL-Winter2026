The largest discount factor such that an optimal agent starting in the initial far-left state ($s_1$) does not swim up the river (does not go RIGHT) is: $discount_factor=0.67, V(s_1)=0.015$ for a weak current, $discount_factor=0.77, V(s_1)=0.022$ for a medium current, and $discount_factor=0.93, V(s_1)=0.068$ for a strong current.

As the discount factor $\gamma \to 0$, the agent focuses almost only on immediate rewards. As the discount factor $\gamma \to 1$, the agent values future rewards almost as much as immediate ones, which also results in higher values in the value functions. 

The agent can get rewards at either $s_1$ ($r=5/1000$) or $s_6$ ($r=1$). Since it can attain rewards by staying at $s_1$, we need a larger discount factor to "push" the agent to go RIGHT. Furthermore, a stronger current increases the probability of the agent being pushed back (the backward state of $s_1$ is still $s_1$). Therefore, when the current becomes stronger, we need an even larger discount factor to motivate the agent to move RIGHT.
